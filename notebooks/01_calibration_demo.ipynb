# Calibração: demonstração sintética

Dados inventados apenas para verificar o fluxo. Não representam metotrexato, vincristina ou qualquer medição da IC. Instale o pacote a partir da raiz com `python -m pip install -e .`.


```python
from dataclasses import asdict
from ic1.analysis import fit_calibration
concentration = [0, 1, 2, 3, 4]
signal = [2, 5, 8, 11, 14]
fit = fit_calibration(concentration, signal)
asdict(fit)
```

```python
assert fit.slope == 3
assert fit.intercept == 2
assert fit.r_squared == 1
assert fit.estimate(9.5) == 2.5
print("Fluxo sintético verificado; não é um resultado experimental.")
```
