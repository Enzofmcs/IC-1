# Fluxo para sinais instrumentais

Este caderno depende dos arquivos originais, não incluídos no repositório. Ajuste os caminhos e a janela conforme o protocolo. Nenhuma planilha é sobrescrita.


```python
from pathlib import Path
from ic1.analysis import consolidate_signals, signal_peak, parse_sample_name, fit_calibration
DATA = Path("../data/raw/nova")  # ajuste para sua pasta
LOWER = None  # definir conforme protocolo
UPPER = None
```

```python
paths = list(DATA.glob("*.xlsx"))
if not paths:
    raise FileNotFoundError("Adicione as planilhas autorizadas em data/raw/nova.")
signals = consolidate_signals(paths)
signals.head()
```

```python
pairs = []
identity = set()
for name in signals.columns[1:]:
    analyte, concentration, unit = parse_sample_name(name)
    identity.add((analyte, unit))
    _, maximum = signal_peak(signals.x, signals[name], LOWER, UPPER)
    pairs.append((concentration, maximum))
if len(identity) != 1:
    raise ValueError("Não misture compostos ou unidades diferentes na mesma curva.")
fit = fit_calibration([x for x, _ in pairs], [y for _, y in pairs])
fit
```

Para espectros CSV, configure explicitamente `x_column`, `y_column` e `encoding="utf-16le"` em `read_signal`. A função não descarta uma coluna de desvio-padrão da fonte e mantém o arquivo original intacto.
