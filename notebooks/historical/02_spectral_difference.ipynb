```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline
```

```python
df = pd.read_excel('ALL SRF & IMT.xlsx')


```

```python
vetor1 = df.iloc[:, 2].values
vetor2 = df.iloc[:, 6].values
vetx = df.iloc[:, 0].values




print(df.columns[2], df.columns[6])
```

```python
def plotsub(column_1, column_2, dataf):
    vetx = dataf.iloc[:, 0].values
    vetor1 = dataf[column_1].values
    vetor2 = dataf[column_2].values   
    name = [column_1, column_2, 'sub']


    x = np.array(sorted(vetx))

    y1 = np.array(vetor1)
    y2 = np.array(vetor2)
    y3 = np.array(vetor2-vetor1)
    i = 0

    nome = []
    # Cria um novo conjunto de x para a interpolação
    xnew = np.linspace(x.min(), x.max(), 500)

    # Para cada vetor y, cria uma spline e plota a linha suave
    for y in [y1, y2, y3]:
        spl = make_interp_spline(x, y, k=3)  # tipo de spline cúbica
        ynew = spl(xnew)
        plt.plot(xnew, ynew, label=name[i])
        i = i + 1



    plt.xlim(188, 400)  # Limita o eixo x entre 180 e 400

    plt.legend(frameon=False, fontsize=10)
    plt.show()
    
```

```python
plotsub('IMT 1uM', 'IMT 1uM SRF 8uM', df)
```
