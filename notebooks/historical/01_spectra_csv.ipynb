```python
# Cria um novo arquivo Excel
wb = openpyxl.Workbook()

# Seleciona a planilha ativa (por padrão, uma nova planilha já é criada)
sheet = wb.active
```

```python
dados = {'wavelenght': save}
newdata = pd.DataFrame(dados)
df = []


for nome_arquivo in os.listdir('arquivoscsv/'):
        arq = pd.read_csv('arquivoscsv/' + nome_arquivo, encoding='UTF-16LE')
        del arq['Std.Dev.']
        arq.columns = ['wave', 'absorb']
        df.append(arq)
        newdata[nome_arquivo] = arq['absorb']


# Valor de referência
valor_referencia = 400

# Encontrar o índice da primeira linha em que a coluna 'wavelenght' é maior ou igual ao valor de referência
indice_referencia = newdata[newdata['wavelenght'] >= valor_referencia].index.min()

# Se houver um índice de referência, manter apenas as linhas até esse índice
if indice_referencia is not None:
    newdata = newdata.loc[:indice_referencia]

with pd.ExcelWriter('newsheet/test1.xlsx', engine='xlsxwriter') as writer:
                # Salva o DataFrame no Excel
                newdata.to_excel(writer, sheet_name='Planilha1', index=False)
```
