```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('sla.xlsx')
```

```python
x = df['Base.Time (s)']
y = df['WE(1).Base.Current (A)']

# Criando o gráfico
plt.figure(figsize=(10,6))
plt.plot(x, y, label='WE(1).Base.Current (A) vs Base.Time (s)')
plt.xlabel('Base.Time (s)')
plt.ylabel('WE(1).Base.Current (A)')
plt.title('Gráfico de Corrente Base vs Tempo Base')
plt.legend()
plt.show()
```

```python
x = df['Potential applied (V)']
y = df['WE(1).δ.Current (A)']

# Criando o gráfico
plt.figure(figsize=(10,6))
plt.plot(x, y, label='WE(1).δ.Current (A) vs Potential applied (V)')
plt.xlabel('Potential applied (V)')
plt.ylabel('WE(1).δ.Current (A)')
plt.title('Gráfico de Corrente δ vs Potencial Aplicado')
plt.legend()
plt.show()
```

```python
# Encontrar o índice do valor máximo
idx_max = df['WE(1).δ.Current (A)'].idxmax()

# Encontrar o valor correspondente no eixo x
x_max = df.loc[idx_max, 'Potential applied (V)']

# Encontrar o valor máximo no eixo y
y_max = df.loc[idx_max, 'WE(1).δ.Current (A)']

print(f'O ponto máximo é em x = {x_max} e y = {y_max}')
```

```python
plt.plot(x, y, label='WE(1).δ.Current (A) vs Potential applied (V)')
plt.plot(x_max, y_max, 'ro')  # adiciona o ponto máximo ao gráfico
plt.xlabel('Potential applied (V)')
plt.ylabel('WE(1).δ.Current (A)')
plt.title('Gráfico de Corrente δ vs Potencial Aplicado')
plt.legend()
plt.show()
```
