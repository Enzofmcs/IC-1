```python
pontos = []

for i in range(1, len(nomes_planilhas)):
    dados_planilha = df.parse(f'Ponto {i}.')
    enz = dados_planilha.iloc[:, 1]
    maxenz = max(enz)


    ant = dados_planilha.iloc[:, 2]
    maxant = max(ant)
    if maxenz < maxant:
        print('erro! o valor máximo de anticorpo é maior que o da enzima')
    pontos.append(maxant - maxenz)
    print(f'enz max {i} = {maxenz}')
    print(f'ant max {i} = {maxant}')

xaxis = dados_planilha.iloc[:, 0]
```

```python
print(pontos)
```

```python
def plot_regressao_linear(x, y):
    # Realizando a regressão linear
    slope, intercept, r_value, p_value, std_err = linregress(x, y)

    # Calculando os valores previstos pela regressão linear
    previstos = slope * np.array(x) + intercept

    # Plotando o gráfico de dispersão e a linha de tendência
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y, color='blue', label='Valores')
    plt.plot(x, previstos, color='red', label='Linha de Tendência')

    # Adicionando legendas e título ao gráfico
    plt.xlabel('Valores de X')
    plt.ylabel('Valores de Y')
    plt.title('Regressão Linear')
    plt.legend()

    # Mostrando o gráfico
    plt.grid(True)
    plt.show()

    # Retornando os coeficientes da regressão
    return slope, intercept, r_value, p_value, std_err
```

```python
dados_planilha = df.parse('curva analitica', header=None)
xaxis = dados_planilha.iloc[:, 0].tolist()
print(dados_planilha)
print(xaxis)
print(sorted(pontos, reverse=True))
res = plot_regressao_linear(xaxis, sorted(pontos, reverse=True))
```
